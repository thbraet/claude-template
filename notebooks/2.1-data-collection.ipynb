## Cross-Dataset Consistency Checks

````python
# PassengerId disjointness
train_ids = set(train["PassengerId"])
test_ids = set(test["PassengerId"])
baseline_ids = set(baseline["PassengerId"])

print("PassengerId ranges:")
print(f"  train:    {train['PassengerId'].min()} – {train['PassengerId'].max()}")
print(f"  test:     {test['PassengerId'].min()} – {test['PassengerId'].max()}")
print(f"  baseline: {baseline['PassengerId'].min()} – {baseline['PassengerId'].max()}")
print()
print(f"Train/test overlap: {train_ids & test_ids or 'None (disjoint)'}")
print(f"Test/baseline alignment: {'Perfect 1:1 match' if test_ids == baseline_ids else 'MISMATCH'}")
print()

# Column consistency between train and test
train_cols = set(train.columns)
test_cols = set(test.columns)
print(f"Columns in train only: {train_cols - test_cols}")
print(f"Columns in test only:  {test_cols - train_cols}")
print()

# Shared columns — dtype consistency
shared_cols = sorted(train_cols & test_cols)
dtype_mismatches = []
for col in shared_cols:
    if train[col].dtype != test[col].dtype:
        dtype_mismatches.append((col, train[col].dtype, test[col].dtype))
print(f"Dtype mismatches on shared columns: {dtype_mismatches or 'None'}")
````

## Key Observations from Initial Profiling

1. **Missingness pattern consistent across train/test:** Age ~20% missing in both sets; Cabin ~77-78% missing in both sets. Imputation strategy must handle both.
2. **Test set has 1 missing Fare:** train.csv has no missing Fare values, but test.csv has 1 (PassengerId 1044). Must handle in preprocessing.
3. **PassengerId ranges are disjoint:** train (1–891), test (892–1309). No overlap, confirming clean train/test split.
4. **Ticket column is messy:** 681 unique values across 891 rows — many tickets are shared by family groups. Potential feature engineering signal (shared ticket = traveling together).
5. **Name column contains titles:** "Mr.", "Mrs.", "Miss.", "Master." embedded in names. Key feature engineering opportunity.
